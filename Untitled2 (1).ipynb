```python
features = pd.read_csv(f"{DATA_DIR}/features.txt", sep=r"\s+", header=None, names=["idx", "feature"])
feature_names = features["feature"].tolist()

# UCI HAR has duplicate feature names -> make them unique to avoid pandas errors
# Manually dedupe feature names (UCI HAR has many duplicate column names)
def dedupe_names(names):
    seen = {}
    result = []
    for name in names:
        if name not in seen:
            seen[name] = 0
            result.append(name)
        else:
            seen[name] += 1
            result.append(f"{name}.{seen[name]}")
    return result

feature_names = dedupe_names(feature_names)
```
